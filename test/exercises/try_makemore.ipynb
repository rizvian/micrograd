```python
import torch
import re
```

```python
f = open ("/home/alirizvi/Documents/only_names.txt","r").read().lower().split('\n')
names = f
```

```python
nameslist = []
for lp in names:
    n1 = lp + "."
    nameslist.append(n1)

names_t = []
for lp in "".join(nameslist):
    names_t.append(lp)

names_d = {}
for lp1, lp2 in zip(names_t[:], names_t[1:]):
    names_d[lp1+lp2] = names_d.get(lp1+lp2, 0) + 1

name_index=[]
for lp in "abcdefghijklmnopqrstuvwxyz.":
    name_index.append(lp)

index_name = {}
cntr = 0
for lp in "abcdefghijklmnopqrstuvwxyz.":
    index_name[lp] = cntr
    cntr += 1
```

```python
t = torch.zeros(27,27)

clp = 0
clp2 = 0
for lp in "abcdefghijklmnopqrstuvwxyz.":
    clp2 = 0
    for lp2 in "abcdefghijklmnopqrstuvwxyz.":
        t[clp,clp2]+=names_d.get(lp+lp2,0)
        clp2 += 1
    clp += 1

tp = t/t.sum(dim=1, keepdim=True)
```

```python
cntr = 0
l="."
named = l    
while True:
    if cntr >= 20:
        break
    i = index_name[l]
    next_l = name_index[tp[i].multinomial(1).item()]
    named += next_l
    l = next_l
    if next_l == ".":
        break
    cntr += 1

named
```
