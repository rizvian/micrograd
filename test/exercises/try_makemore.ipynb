```python
import torch
```

```python
names = open ("/home/alirizvi/Documents/only_names.txt","r").read().lower().split('\n')
```

```python
nameslist = []
for lp in names:
    n1 = "." + lp 
    nameslist.append(n1)

names_t = []
for lp in "".join(nameslist):
    names_t.append(lp)

names_d = {}
for lp1, lp2 in zip(names_t[:], names_t[1:]):
    names_d[lp1+lp2] = names_d.get(lp1+lp2, 0) + 1

name_index=[]
for lp in "abcdefghijklmnopqrstuvwxyz.":
    name_index.append(lp)

index_name = {}
cntr = 0
for lp in "abcdefghijklmnopqrstuvwxyz.":
    index_name[lp] = cntr
    cntr += 1
```

```python
t = torch.zeros(27,27)

clp = 0
clp2 = 0
for lp in "abcdefghijklmnopqrstuvwxyz.":
    clp2 = 0
    for lp2 in "abcdefghijklmnopqrstuvwxyz.":
        t[clp,clp2]=names_d.get(lp+lp2,0)
        clp2 += 1
    clp += 1
    
# dim=0 sums up all the columns (think of slider moving in x axis)
# dim=1 sums up all the rows (think of slider moving in y axis)
# keepdim keeps a 2d matrix e.g. [len(t),1] otherwise ...
# len(t) will automatically move towards the right side ...
# during broadcasting --> [x,y] + [_,len(t)] ...
# instead of [x,y]+[len(t),1]
t_rowsum = t.sum(dim=1, keepdim=True)
tp = t/t_rowsum
tp += 10**-10 # prevent divide by zero


# uniform prob for all combinations - untrained model
t_dummy_rowsum = torch.ones(27,1)
tp_dummy = torch.ones(27,27)/t_dummy_rowsum
tp_dummy += 10**-10 # prevent divide by zero

cnt=0
names_dp = {}
for lp1, lp2 in zip(name_index[:], name_index[1:]):
    names_d[lp1+lp2] = names_d.get(lp1+lp2, 0) + 1
    cnt += 1 
```

```python
tp_run, tp_rowrun = tp, t_rowsum
# tp_run, tp_rowrun = tp_dummy, t_dummy_rowsum

cntr = 0
l="."
named = l

while True:
    if cntr >= 15:
        break
    i = index_name[l]
    next_l = name_index[tp_run[i].multinomial(1).item()]
    if cntr < 3 and next_l==".":
        continue
    named += next_l
    l = next_l
    if next_l == ".":
        break
    cntr += 1

named = ".jemmaq."
print(named)

prob_array = torch.zeros(len(named)-1)
for lp in range(len(named)-1):
    # print(named[lp], named[lp+1])
    # print(index_name[named[lp]],index_name[named[lp+1]])
    # print(tp[index_name[named[lp]],index_name[named[lp+1]]].item())
    prob_array[lp]=tp[index_name[named[lp]],index_name[named[lp+1]]].item()

print(prob_array)
print(-torch.log(prob_array).sum()) #negative log likelihood (NLL)
print(-torch.log(prob_array).sum()/len(named)) #negative average log likelihood (NALL)
```
